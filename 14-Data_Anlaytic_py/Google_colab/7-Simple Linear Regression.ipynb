Distribution plot 1

```python
import seaborn as sns 
import numpy as np 
np.random.seed(0)
x =    np.random.rand(100)
ax_1 = sns.displot(x)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

ax_2 = sns.displot(x=x, rug=True, kind="kde")
plt.show()
```

```python
sns.rugplot(x=x)
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.stats import norm

sns.histplot(x=x, stat="density", kde=False)

# Fit normal distribution
mu, std = norm.fit(x)
xmin, xmax = plt.xlim()
import numpy as np
x_axis = np.linspace(xmin, xmax, 100)

sns.lineplot(x=x_axis, y=norm.pdf(x_axis, mu, std))

plt.show()
```
