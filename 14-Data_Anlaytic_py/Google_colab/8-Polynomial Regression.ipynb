```python
import numpy as np

x = np.random.randn(100)
y = np.random.randn(100)

f = np.polyfit(x, y, 2)  # degree-2 polynomial fit
p = np.poly1d(f)

print(f)       # polynomial coefficients
print(p)       # polynomial equation
```

```python
from sklearn.preprocessing import PolynomialFeatures
import seaborn as sns

flights = sns.load_dataset("flights")

pr = PolynomialFeatures(degree=3, include_bias=False)
x_poly = pr.fit_transform(flights[["year", "passengers"]])

print(x_poly[:5])
```
