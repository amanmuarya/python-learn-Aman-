```python
import pandas as pd

ab = pd.DataFrame({
    "Marks": [12, 18, 25, 32, 45, 52, 67, 75, 88, 95]
})

ab["Category"] = pd.cut(
    ab["Marks"],
    bins=[0, 30, 60, 100],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)

print(ab)
```

```python
import numpy as np
import pandas as pd

path = "E:/Excel_course/aman_45.xlsx"
ab = pd.read_excel(path)

# Create bin boundaries
bins = np.linspace(ab["Sub1"].min(), ab["Sub1"].max(), 4)

# Create bin labels
group_names = ["Low", "Medium", "High"]

# Divide values into bins
ab["Sub1_binned"] = pd.cut(
    ab["Sub1"],
    bins=bins,
    labels=group_names,
    include_lowest=True
)

print(ab)
```
